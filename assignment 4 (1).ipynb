```python
import numpy as np

arr_1 = np.array([[1, 2, 3, 4],
                  [5, 6, 7, 8]])

arr_2 = np.array([[9, 8, 7, 4],
                  [3, 12, 4, 8]])

arr_3 = arr_1 + arr_2

print("First Array:")
print(arr_1)

print("Second Array:")
print(arr_2)

print("Addition of Two Arrays:")
print(arr_3)
```
