```python
# Creating a List
students_list = ["Aditi", "Shaurya", "Rishi", "Sneha", "Riya"]

print("Students List:", students_list)

# Adding an element
students_list.append("Rohan")
print("After adding Rohan:", students_list)

# Removing an element
students_list.remove("Aditi")
print("After removing Aditi:", students_list)

# Removing an element using pop()
students_list = ["Aditi", "Shaurya", "Rishi", "Sneha", "Riya"]
students_list.pop(1)
print("After using pop():", students_list)

# Deleting an element using del
del students_list[3]
print("After deleting an element:", students_list)

# Updating an element
students_list[2] = "Shreya"
print("After updating an element:", students_list)


# Creating a Tuple
students_tuple = ("Aditi", "Shaurya", "Rishi", "Sneha", "Riya")

print("\nStudents Tuple:", students_tuple)

# Adding elements to a tuple using a list
y = list(students_tuple)
y.append("Rohan")
y.append("Himesh")
y.append("Karan")

students_tuple = tuple(y)
print("After adding elements:", students_tuple)

# Adding another element
students_tuple += ("Riley",)
print("After adding Riley:", students_tuple)

# Removing elements from a tuple using a list
y = list(students_tuple)
y.remove("Aditi")
y.remove("Riya")

students_tuple = tuple(y)
print("After removing elements:", students_tuple)


# Creating a Dictionary
students_dict = {
    1: "Aditi",
    2: "Shaurya",
    3: "Rishi",
    4: "Sneha",
    5: "Riya"
}

print("\nStudents Dictionary:", students_dict)

# Adding an element
students_dict[6] = "Rohan"
print("After adding Rohan:", students_dict)

# Deleting an element
del students_dict[3]
print("After deleting Rishi:", students_dict)

# Updating an element
students_dict[2] = "Shreya"
print("After updating Shreya:", students_dict)
```
